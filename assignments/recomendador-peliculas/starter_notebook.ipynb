## Exploración rápida

Antes de modelar, vale la pena mirar cuántos usuarios/películas hay,
qué tan dispersa es la matriz de utilidad, y si hay usuarios o
películas con muy pocas interacciones (posible cold-start).

````python
n_usuarios = train["user_id"].nunique()
n_peliculas = movies["item_id"].nunique()
sparsity = 1 - len(train) / (n_usuarios * n_peliculas)

print(f"Usuarios en train: {n_usuarios}")
print(f"Películas en catálogo: {n_peliculas}")
print(f"Sparsity aproximada de train: {sparsity:.4f}")
````
