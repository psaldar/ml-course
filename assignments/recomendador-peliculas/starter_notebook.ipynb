## 5. Entrega

```python
def construir_submission(ids, predicciones, ruta="mi_prediccion.csv"):
    """Arma el CSV en el formato que espera el leaderboard y valida el formato."""
    sub = pd.DataFrame({"id": ids, "rating": predicciones})
    if len(sub) != 20381:
        raise ValueError(f"Se esperaban 20381 filas, hay {len(sub)}")
    if sub["id"].duplicated().any():
        raise ValueError("Hay ids duplicados")
    if sub["rating"].isna().any():
        raise ValueError("Hay predicciones vacías")
    if not sub["rating"].between(1, 5).all():
        raise ValueError("Las calificaciones deben quedar entre 1 y 5")
    sub.to_csv(ruta, index=False)
    return sub


# construir_submission(test["id"], mis_predicciones)
```
