## 4. Entrega

```python
def construir_submission(ids, predicciones, ruta="mi_prediccion.csv"):
    """Arma el CSV en el formato que espera el leaderboard y valida el formato."""
    sub = pd.DataFrame({"id": ids, "cnt": predicciones})
    if len(sub) != 2168:
        raise ValueError(f"Se esperaban 2168 filas, hay {len(sub)}")
    if sub["id"].duplicated().any():
        raise ValueError("Hay ids duplicados")
    if (sub["cnt"] < 0).any():
        raise ValueError("cnt es un conteo: no puede ser negativo")
    sub.to_csv(ruta, index=False)
    return sub


# construir_submission(test["id"], mis_predicciones)
```
